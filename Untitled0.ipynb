```python
import numpy as np
import pandas as pd

# 1. Create a dummy raw dataset with missing values and categories
raw_data = {
    "Student_ID": [1, 2, 3, 4, 5],
    "Age": [18, 19, np.nan, 20, 19],
    "Branch": ["CS", "AI", "CS", np.nan, "AI"],
    "Exam_Score": [85.0, np.nan, 90.0, 78.0, 92.0],
    "Passed": ["Yes", "No", "Yes", "No", "Yes"],
}

df = pd.DataFrame(raw_data)
print("--- Original Raw Dataset ---")
print(df)
print("\n" + "=" * 40 + "\n")

# 2. Handling Missing Values (Data Cleaning)
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Exam_Score"] = df["Exam_Score"].fillna(df["Exam_Score"].mean())
df["Branch"] = df["Branch"].fillna(df["Branch"].mode()[0])

# 3. Encoding Categorical Variables
df["Passed"] = df["Passed"].map({"Yes": 1, "No": 0})
df = pd.get_dummies(df, columns=["Branch"], dtype=int)

# 4. Normalization (Scaling Exam Scores between 0 and 1)
min_score = df["Exam_Score"].min()
max_score = df["Exam_Score"].max()
df["Normalized_Score"] = (df["Exam_Score"] - min_score) / (
    max_score - min_score
)

print("--- Cleaned and Preprocessed Dataset ---")
print(df)
```
